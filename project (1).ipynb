```python
students = []
def error_handling(message):
    while True:
        try:
            return int(input(message))
        except ValueError:
            print("Invalid input! Please enter a number.")
def add_student():
    name = input("Enter Student name: ")
    student_id = error_handling("Enter your ID: ")
    age = error_handling("Enter your age: ")
    grade_math = error_handling("Enter your grade in Math 3: ")
    grade_circut = error_handling("Enter your grade in Circut 1: ")
    grade_electronic = error_handling("Enter your grade in Electronic 1: ")
    student_information = {
        "name": name,
        "ID": student_id,
        "age": age,
        "Grades": [grade_math, grade_circut, grade_electronic],
    }
    students.append(student_information)
def search_student():
    search_id = error_handling("Enter the Student ID: ")
    for student in students:
        if student["ID"] == search_id:
            print(student)
            return
    print("This ID Not Exist !")
def view_student():
    if len(students) == 0:
        print("No students yet.")
        return
    for student in students:
        print(student)
def Update_Grades():
    search_id = error_handling("Enter student ID: ")
    for student in students:
        if student["ID"] == search_id:
            student["Grades"][0] = error_handling("New grade in Math 3: ")
            student["Grades"][1] = error_handling("New grade in Circuit 1: ")
            student["Grades"][2] = error_handling("New grade in Electronic 1: ")
            print("Grades updated!")
            return
    print("This ID does not exist!")
def calculate_avg():
    search_id = error_handling("Enter The Student ID: ")
    for student in students:
        if student["ID"] == search_id:
            avg = sum(student["Grades"]) / len(student["Grades"])
            print("Avradge Grades : " + str(avg))
            if avg >= 90:
                print("Excellent")
            elif avg >= 80:
                print("Very Good")
            elif avg >= 70:
                print("Good")
            elif avg >= 50:
                print("pass")
            else:
                print("Fail")
            return
    print("This ID Not Exist !")
def show_top_student():
    if len(students) == 0:
        print("No students yet.")
        return
    top_student = students[0]
    top_average = sum(top_student["Grades"]) / len(top_student["Grades"])
    for student in students:
        average = sum(student["Grades"]) / len(student["Grades"])
        if average > top_average:
            top_average = average
            top_student = student
    print("Top student:", top_student["name"])
    print("Average:", round(top_average, 2))
def delete_student():
    delete_id = error_handling("Enter student ID to delete: ")
    for student in students:
        if student["ID"] == delete_id:
            students.remove(student)
            print("Student deleted successfully!")
            return
    print("This ID does not exist!")
def show_statistics():
    passing_avg = 50
    if len(students) == 0:
        print("No students yet.")
        return
    first_avg = sum(students[0]["Grades"]) / len(students[0]["Grades"])
    highest = first_avg
    lowest = first_avg
    total_of_averages = 0
    passing = 0
    failing = 0
    for student in students:
        average = sum(student["Grades"]) / len(student["Grades"])
        total_of_averages = total_of_averages + average
        if average > highest:
            highest = average
        if average < lowest:
            lowest = average
        if average >= passing_avg:
            passing = passing + 1
        else:
            failing = failing + 1
    class_average = total_of_averages / len(students)
    print("Total students:", len(students))
    print("Highest average:", round(highest, 2))
    print("Lowest average:", round(lowest, 2))
    print("Class average:", round(class_average, 2))
    print("Passing students:", passing)
    print("Failing students:", failing)
def show_menu():
    while True:
        print("=================================")
        print ("   STUDENT MANAGEMENT SYSTEM")
        print("=================================")
        print("1. Add a New Student")
        print("2. View All Students")
        print("3. Search for a Student")
        print("4. Update Student Grades")
        print("5. Calculate Student Average")
        print("6. Show the Top Student")
        print("7. Delete a Student")
        print("8. Show Statistics")
        print("9. Exit")
        operation = error_handling("Enter your choice (1-9): ")
        if operation == 1:
            add_student()
        elif operation == 2:
            view_student()
        elif operation == 3:
            search_student()
        elif operation == 4:
            Update_Grades()
        elif operation == 5:
            calculate_avg()
        elif operation == 6:
            show_top_student()
        elif operation == 7:
            delete_student()
        elif operation == 8:
            show_statistics()
        elif operation == 9:
            print("Exit Program")
            break
        else:
            print("This Number Invalid !")
        print()
show_menu()
```
